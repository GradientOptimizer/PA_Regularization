### StandardScaler

The StandardScaler transforms data such that each feature has:

- A mean of 0 $(\mu = 0)$

- A standard deviation of 1 $(\sigma = 1)$

$$
X_\text{scaled} = \frac{X - \mu}{\sigma}
$$​

where 

$X$ is the original value

$\mu$ is the mean of the feature

$\sigma$ is the standard deviation.

Suppose we set `scaler = StandardScaler()`

StandardScaler has many methods such as 

`fit(X, y=None, sample_weight=None)` which computes the mean and standard deviation to be used for later scaling.

`fit_transform(X, y=None, **fit_params)` which
fit to data, then transform it. Fits transformer to X and y with optional parameters fit_params and returns a transformed version of X.

`inverse_transform(X, copy=None)` which scales back the data to the original representation.

`transform(X, copy=None)` which perform standardization by centering and scaling.

source: https://scikit-learn.org/stable/modules/generated/sklearn.preprocessing.StandardScaler.html#sklearn.preprocessing.StandardScaler.transform

`fit_transform()` computes the mean and standard deviation of the training data, 
then uses these fitted or learned statistics to standardize the data to 
a mean of 0 and a standard deviation of 1. 
Meanwhile, `transform()` uses the previously fitted 
statistics to standardize new data without recomputing them.


This ensures that both the training and test sets are standardized using the same mean and standard deviation, maintaining a consistent feature scale and preventing the test set's statistical information from influencing the preprocessing procedure.

| **Method** | **Learns Parameters** | **Transforms Data** | **Returns** |
| :---: | :---: | :---: | :---: |
| `fit()` | Yes | No | Fitted estimator |
| `transform()` | No | Yes | Transformed data |
| `fit_transform()` | Yes | Yes | Transformed data |

the fitted statistics from `fit_transform()` is passed onto `scaler`.

source: https://www.geeksforgeeks.org/python/what-is-the-difference-between-transform-and-fit_transform-in-sklearn-python/

```python
from sklearn.preprocessing import StandardScaler 

scaler = StandardScaler()

# passes the learned statistics to scaler
X_train_scaled = scaler.fit_transform(X_train)  

# the learned statistics from scaler is used to transform X_test
X_test_scaled = scaler.transform(X_test)
```

```python
display(type(X_train_scaled))
display(type(X_test_scaled))

# scaler returns a numpy array instead of DataFrame
# to use a .hist() method easily, we need to convert into DataFrame
```

```python
X_train_scaled = pd.DataFrame(
    X_train_scaled,
    columns = X_train.columns,  # column/feature names
    index = X_train.index  # row indexes  
)

X_test_scaled = pd.DataFrame(
    X_test_scaled,
    columns = X_test.columns,  
    index = X_test.index   
)

display(type(X_train_scaled))
display(type(X_test_scaled))
```

## 2. Hyperparameter Tuning with Cross-Validation (15 pts):

- Instantiate an `ElasticNet(random_state=42)` model.

- Set up a `GridSearchCV` pipeline testing the following grid:

`alpha`: [0.01, 0.1, 1.0, 10.0, 100.0]

`l1_ratio`: [0.1, 0.3, 0.5, 0.7, 0.9]

- Perform 5-fold cross-validation (`cv=5`) using negative mean squared error
(`scoring='neg_mean_squared_error'`) or R² (`scoring='r2'`) as the scoring metric.

```python
from sklearn.linear_model import ElasticNet

baseline_model = ElasticNet(random_state=42)
baseline_model.fit(X_train_scaled, y_train)  # Fits the untuned model
```

```python
from sklearn.model_selection import GridSearchCV 

my_param_grid = {
    'alpha': [0.01, 0.1, 1.0, 10.0, 100.0],
    'l1_ratio': [0.1, 0.3, 0.5, 0.7, 0.9]
}

grid = GridSearchCV(
    estimator = baseline_model,
    param_grid = my_param_grid,
    cv = 5,
    scoring = ['neg_mean_squared_error', 'r2'],
    refit = 'r2'
)

grid.fit(X_train_scaled, y_train)  # Fits the tuned model 
```

## 3. Evaluation & Coefficient Analysis (15 pts):

Extract the best estimator from the grid search and display the optimal hyperparameters
(`best_params_`).

```python
best_score = grid.best_score_
best_parameters = grid.best_params_
best_model = grid.best_estimator_  

print(f"Best r2: {best_score:.4f}")
print(f"Best Parameters: {best_parameters}")
print(f"Best Estimator (Model): {best_model}")
```

Predict on the unseen scaled test set (`X_test_scaled`) and evaluate performance by calculating
Root Mean Squared Error (RMSE) and R² score (`r2_score`)

```python
from sklearn.metrics import root_mean_squared_error
from sklearn.metrics import r2_score

y_pred_untuned = baseline_model.predict(X_test_scaled)  # Baseline ElasticNet
y_pred_tuned = best_model.predict(X_test_scaled)  # Tuned ElasticNet

rmse_untuned = root_mean_squared_error(y_test, y_pred_untuned)
r2_untuned = r2_score(y_test, y_pred_untuned)

rmse_tuned = root_mean_squared_error(y_test, y_pred_tuned)
r2_tuned = r2_score(y_test, y_pred_tuned)

print("Unoptimized ElasticNet predictive metrics:")
print(f"Root Mean Squared Error: {rmse_untuned:.4f}")
print(f"r2 score: {r2_untuned:.4f}")

print("\nTuned ElasticNet predictive metrics:")
print(f"Root Mean Squared Error: {rmse_tuned:.4f}")
print(f"r2 score: {r2_tuned:.4f}")
```

Extract the feature coefficients (`best_model.coef_`), map them to feature names, and display
the total number of features zeroed out (coefficients equal to 0)

```python
baseline_coef = pd.Series(data = baseline_model.coef_, index = X_train.columns)
best_coef = pd.Series(data = best_model.coef_, index = X_train.columns)
```

```python
print("Coefficients of unoptimized ElasticNet")
display(baseline_coef.to_frame().style)

print("Coefficients of Tuned ElasticNet")
display(best_coef.to_frame().style)
```

```python
"""
+= is equivalent to x = x + 1
=+ is equivalent to x = 1 or x = +1   
"""

baseline_zeros = 0

for i in range(0, len(baseline_coef)):     
    if baseline_coef.iloc[i] == 0:
        baseline_zeros += 1 
        
tuned_zeros = 0 

for i in range(0, len(best_coef)):
    if best_coef.iloc[i] == 0:
        tuned_zeros += 1
        
print(f"Number of coefficients equal to zero for unoptimized ElasticNet: {baseline_zeros}")    
print(f"Number of coefficients equal to zero for Tuned ElasticNet: {tuned_zeros}") 
```
